# Experiment 7 (Valid Variable Name)

```python
%%writefile exp7.y
%{
#include <stdio.h>
#include <stdlib.h>
void yyerror(const char *s);
int yylex();
%}

%token ID

%%
stmt : ID   { printf("Valid Variable Name\n"); }
     ;
%%

void yyerror(const char *s) {
    printf("Invalid Variable Name\n");
}

int main() {
    printf("Enter a variable name: ");
    yyparse();
    return 0;
}
```

```python
%%writefile exp7.l
%{
#include "y.tab.h"
%}

%%
[a-zA-Z][a-zA-Z0-9]*   { return ID; }
\n                     { return 0; }
.                      { return yytext[0]; }
%%
```

```python
%%shell
yacc -d exp7.y
lex exp7.l
cc y.tab.c lex.yy.c -ll
./a.out
```

# Experiment 8 (a^n b)

```python
%%writefile exp8.y

%{
#include<stdio.h>
#include<stdlib.h>
int yylex();
void yyerror(const char *s);
%}
%token A B
%%
input:	s {printf("Successful Grammar\n"); exit(0);}
	s: A s1 B| B
	s1: ; | A s1
%%

int main()
{
	printf("\nEnter A String: ");
	yyparse();
	return 0;
}
void yyerror(const char *s)
{
	printf("\nError: Invalid String\n");
	exit(0);
}
```

```python
%%writefile exp8.l

%{
#include "y.tab.h"
%}
%%
a       {return A;}
b       {return B;}
[ \t\n]  ; /* ignore whitespace */
.       {return yytext[0];}
%%
```

```python
%%shell

yacc -d exp8.y
lex exp8.l
cc y.tab.c lex.yy.c -ll
./a.out
```

# Experiment 5 (Evaluating Arithmetic Expressions)

```python
%%writefile exp5.y

%{
#include <stdio.h>
#include <stdlib.h>

int yylex(void);
int yyerror();
%}

%token num
%left '+' '-'
%left '*' '/'

%%

input: exp { printf("%d\n", $$); exit(0); }

exp: exp '+' exp { $$ = $1 + $3; }
   | exp '-' exp { $$ = $1 - $3; }
   | exp '*' exp { $$ = $1 * $3; }
   | exp '/' exp {
        if ($3 == 0) {
            printf("Divide by Zero. Invalid expression.\n");
            exit(0);
        } else {
            $$ = $1 / $3;
        }
     }
   | '(' exp ')' { $$ = $2; }
   | num { $$ = $1; }
   ;

%%

int yyerror()
{
    printf("Error. Invalid Expression.\n");
    exit(0);
}

int main()
{
    printf("Enter an expression:\n");
    yyparse();
}
```

```python
%%writefile exp5.l

%{
#include "y.tab.h"
int extern yylval;
%}

%%

[0-9]+        { yylval = atoi(yytext); return num; }
[\+\-\*\/]    { return yytext[0]; }
[)]           { return yytext[0]; }
[(]           { return yytext[0]; }
.             { ; }
\n            { return 0; }

%%
```

```python
%%shell

yacc -d exp5.y
lex exp5.l
cc y.tab.c lex.yy.c -ll
./a.out
```
